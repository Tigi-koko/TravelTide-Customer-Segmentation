# Data Extraction & Cleaning with **SQL**

```python
sql_query = """

-- Q1: Select sessions after Jan 4, 2023
-- We filter to recent activity, as per Elena’s guidance,
-- to capture current, active user behaviors.
WITH sessions_2023 AS (
    SELECT *
    FROM sessions
    WHERE session_start > '2023-01-04'
),

-- Q2: Filter users with >7 sessions
-- Business rule: defines "active cohort".
-- Rationale: Users below this threshold are too infrequent
-- to be reliable targets for perks/segmentation.
filtered_users AS (
    SELECT user_id
    FROM sessions_2023
    GROUP BY user_id
    HAVING COUNT(*) > 7
),

-- Q3: Build session base (sessions + users + trips)
-- COALESCE/CASE are used here:
-- - Nights: negative values → reset to 0 (stay length cannot be negative).
-- - Null numeric fields (discounts, price, rooms) → 0 (treat as “no discount/no price”).
-- - Rooms: default to 1 to avoid multiplying by NULL.
session_base AS (
    SELECT
        s.session_id, s.user_id, s.trip_id,
        s.session_start, s.session_end,
        EXTRACT(EPOCH FROM s.session_end - s.session_start) AS session_duration,
        s.page_clicks,
        s.flight_discount, s.flight_discount_amount,
        s.hotel_discount, s.hotel_discount_amount,
        s.flight_booked, s.hotel_booked,
        s.cancellation,

        -- Demographics
        u.birthdate, u.gender, u.married, u.has_children,
        u.home_country, u.home_city, u.home_airport,
        u.home_airport_lat, u.home_airport_lon, u.sign_up_date,

        -- Flight details
        f.origin_airport, f.destination_airport, f.seats,
        f.return_flight_booked, f.departure_time, f.return_time,
        f.checked_bags, f.trip_airline,
        f.destination_airport_lat, f.destination_airport_lon,
        f.base_fare_usd,

        -- Hotel details
        h.hotel_name,
        CASE WHEN h.nights < 0 THEN 0 ELSE COALESCE(h.nights, 0) END AS nights,
        COALESCE(h.rooms, 1) AS rooms,
        h.check_in_time, h.check_out_time,
        COALESCE(h.hotel_per_room_usd, 0) AS hotel_price_per_room_night_usd

    FROM sessions_2023 s
    LEFT JOIN users u ON s.user_id = u.user_id
    LEFT JOIN flights f ON s.trip_id = f.trip_id
    LEFT JOIN hotels h ON s.trip_id = h.trip_id
    WHERE s.user_id IN (SELECT user_id FROM filtered_users)
),

-- Q4: Aggregate to user level
-- All metrics are computed at the user level.
-- COALESCE ensures no NULLs leak through.
-- Business impact: makes the dataset ready for clustering & Tableau.
user_level_agg AS (
    SELECT
        s.user_id,

        -- Browsing behavior
        COUNT(DISTINCT s.session_id) AS num_sessions,
        AVG(COALESCE(s.session_duration, 0)) AS avg_session_duration,
        SUM(COALESCE(s.page_clicks, 0)) AS num_clicks,

        -- Booking/cancellation
        COALESCE(SUM(CASE WHEN s.cancellation THEN 1 ELSE 0 END), 0) AS sessions_with_cancellation,
        COALESCE(SUM(CASE WHEN s.flight_discount THEN 1 ELSE 0 END), 0) AS sessions_with_flight_discount,
        COALESCE(SUM(CASE WHEN s.hotel_discount THEN 1 ELSE 0 END), 0) AS sessions_with_hotel_discount,
        COALESCE(SUM(CASE WHEN s.flight_booked THEN 1 ELSE 0 END), 0) AS sessions_with_flight_booked,
        COALESCE(SUM(CASE WHEN s.hotel_booked THEN 1 ELSE 0 END), 0) AS sessions_with_hotel_booked,

        -- Travel details
        COALESCE(AVG(COALESCE(f.checked_bags, 0)), 0) AS avg_bags,
        COALESCE(AVG(COALESCE(s.flight_discount_amount, 0)), 0) AS avg_flight_discount_amount,
        COALESCE(AVG(COALESCE(s.hotel_discount_amount, 0)), 0) AS avg_hotel_discount_amount,

        COUNT(DISTINCT s.trip_id) AS num_trips,
        COALESCE(SUM(
            CASE
                WHEN s.flight_booked AND f.return_flight_booked THEN 2
                WHEN s.flight_booked THEN 1
                ELSE 0
            END
        ), 0) AS num_flights,

        COALESCE(AVG(
            CASE
                WHEN f.departure_time IS NOT NULL AND s.session_end IS NOT NULL
                THEN EXTRACT(DAY FROM f.departure_time - s.session_end)
                ELSE 0
            END
        ), 0) AS time_after_booking_days,

        COALESCE(AVG(COALESCE(f.seats, 0)), 0) AS avg_seats_per_flight,
        COALESCE(AVG(COALESCE(h.nights, 0)), 0) AS avg_nights_per_hotel_stay,

        -- Q5: Distance calculation → moved to Python (Haversine)
        -- Rationale: easier to compute with vectorized NumPy.

        -- Q6: Exploration (new destinations)
        COALESCE(SUM(CASE WHEN s.flight_booked AND f.destination_airport != u.home_airport THEN 1 ELSE 0 END), 0)
        AS num_flight_to_new_destination,

        -- Q7: Seasonality
        COALESCE(SUM(CASE WHEN s.flight_booked AND EXTRACT(MONTH FROM f.departure_time) IN (6,7,8) THEN 1 ELSE 0 END), 0)
        AS num_summer_flights,
        COALESCE(SUM(CASE WHEN s.hotel_booked AND EXTRACT(MONTH FROM h.check_in_time) IN (12,1,2) THEN 1 ELSE 0 END), 0)
        AS num_winter_hotel_stays,
        COALESCE(AVG(h.hotel_per_room_usd), 0) AS hotel_price_per_room_night_usd,
        COALESCE(AVG(h.nights), 0) AS avg_nights,
        COALESCE(AVG(h.rooms), 0) AS avg_rooms,


        -- Q8: Demographics
        COALESCE(MAX(u.birthdate), '1900-01-01') AS birthdate,
        COALESCE(MAX(u.gender), 'Unknown') AS gender,
        COALESCE(BOOL_OR(u.married), FALSE) AS married,
        COALESCE(BOOL_OR(u.has_children), FALSE) AS has_children,
        COALESCE(MAX(u.home_country), 'Unknown') AS home_country,
        COALESCE(MAX(u.home_city), 'Unknown') AS home_city,
        COALESCE(MAX(u.home_airport), 'Unknown') AS home_airport,
        COALESCE(MAX(u.home_airport_lat), 0) AS home_airport_lat,
        COALESCE(MAX(u.home_airport_lon), 0) AS home_airport_lon,
        COALESCE(MAX(f.destination_airport_lat), 0) AS avg_destination_airport_lat,
        COALESCE(MAX(f.destination_airport_lon), 0) AS avg_destination_airport_lon,
        COALESCE(MAX(u.sign_up_date), '1900-01-01') AS sign_up_date,
        COALESCE(AVG(f.base_fare_usd), 0) AS avg_base_fare_usd


    FROM session_base s
    LEFT JOIN users u ON s.user_id = u.user_id
    LEFT JOIN flights f ON s.trip_id = f.trip_id
    LEFT JOIN hotels h ON s.trip_id = h.trip_id
    GROUP BY s.user_id
)

-- Final export
SELECT * FROM user_level_agg;
"""
```

# **Feature Engineering**

```python
# Importing all the nessaarry Libraries for Feauture Engineering
from datetime import datetime
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans, DBSCAN
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from fpdf import FPDF
from math import radians, sin, cos, sqrt, atan2
```

```python
#Convert date columns to datetime format
final_df['sign_up_date'] = pd.to_datetime(final_df['sign_up_date'])
final_df['birthdate'] = pd.to_datetime(final_df['birthdate'])
```

```python
#Calculate age in years and customer age in months
final_df['age'] = (pd.to_datetime('2025-07-01') - final_df['birthdate']).dt.days // 365
final_df['customer_age_months'] = (pd.to_datetime('2025-07-01') - final_df['sign_up_date']).dt.days // 30
```

```python
# total_hotel_revenue_usd already exists from SQL
final_df['total_flight_fare_usd'] = final_df['avg_base_fare_usd'].fillna(0) * final_df['num_flights'].fillna(0)
```

```python
import numpy as np

#Proportion of sessions with flight and hotel discounts
final_df['flight_discount_proportion'] = final_df['sessions_with_flight_discount'] / final_df['num_sessions'].replace(0, np.nan)
final_df['hotel_discount_proportion'] = final_df['sessions_with_hotel_discount'] / final_df['num_sessions'].replace(0, np.nan)

# Fill NaNs with 0 after division
final_df['flight_discount_proportion'] = final_df['flight_discount_proportion'].fillna(0)
final_df['hotel_discount_proportion'] = final_df['hotel_discount_proportion'].fillna(0)
```

```python
#Average discount score combining flight and hotel
final_df['avg_discount_score'] = final_df[['avg_flight_discount_amount', 'avg_hotel_discount_amount']].mean(axis=1)
```

```python
# Bargain Index captures both discount usage and spend
# → helps identify "deal-seekers" vs. "full-fare" customers
final_df['bargain_index'] = (
    final_df['flight_discount_proportion'].fillna(0) *
    final_df['avg_flight_discount_amount'].fillna(0) *
    final_df['total_flight_fare_usd'].fillna(0)
)
```

```python
# Flights: base fare × number of flights
final_df['total_flight_fare_usd'] = (
    final_df['avg_base_fare_usd'].fillna(0) *
    final_df['num_flights'].fillna(0)
)
```

```python
# Hotels: price per room-night × avg nights per stay × avg rooms
final_df['total_hotel_revenue_usd'] = (
    final_df['hotel_price_per_room_night_usd'].fillna(0) *
    final_df['avg_nights_per_hotel_stay'].fillna(0) *
    final_df['avg_rooms'].fillna(0)
)
```

```python
# Total revenue = flights + hotels
final_df['total_revenue'] = (
    final_df['total_flight_fare_usd'] +
    final_df['total_hotel_revenue_usd']
)
```

```python
# --- Distance between home and destination airports ---

def haversine(lat1, lon1, lat2, lon2):
    R = 6371  # Earth radius in km
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat/2)**2 + np.cos(lat1)*np.cos(lat2)*np.sin(dlon/2)**2
    c = 2 * np.arcsin(np.sqrt(a))
    return R * c

# Apply haversine using your dataframe columns
final_df['km_distance'] = final_df.apply(
    lambda row: haversine(
        row['home_airport_lat'], row['home_airport_lon'],
        row['avg_destination_airport_lat'], row['avg_destination_airport_lon']
    ),
    axis=1
)
```

```python
#Exploration Index = % of flights to new destinations
final_df['exploration_index'] = final_df['num_flight_to_new_destination'] / final_df['num_flights'].replace(0, np.nan)
```

# **Clustering with KMeans**

* KMeans is an unsupervised learning method that groups customers into K clusters based on behavior similarity.It works by minimizing the distance between points and their assigned cluster centers.I chose it for its simplicity, speed, and effectiveness in customer segmentation.




```python
scaler = StandardScaler()
# Select the features for clustering and assign them to X
X = features_df[feature_cols]
X_scaled = scaler.fit_transform(X)
```

```python
## KMeans business-friendly fixed groups, fast.
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
kmeans.fit(X_scaled)
```

```python
features_df['cluster'] = kmeans.labels_

segment_names = {
    0: 'Bargain Seekers',
    1: 'Family Travelers',
    2: 'Loyal Explorers',
    3: 'Infrequent Users'
}
features_df['segment'] = features_df['cluster'].map(segment_names)
```

# **DBSCAN for noise handling**

* Why DBSCAN? Unlike KMeans, DBSCAN does not require us to predefine the number of clusters.
It groups together points that are close in high-density areas and labels outliers as noise (-1).This method is helpful to detect unusually behaving users or small niche groups.However, DBSCAN was sensitive to parameter tuning and resulted in many points labeled as noise.In our case, it confirmed the core clusters found by KMeans but added noise points that may represent rare behaviors.





```python
# DBSCAN - finds niche/outliers, no need to pre-set K.
dbscan = DBSCAN(eps=0.6, min_samples=6)
dbscan_labels = dbscan.fit_predict(X_scaled)
features_df['dbscan_cluster'] = dbscan_labels

print(" DBSCAN label counts (cluster vs noise)")
print(features_df['dbscan_cluster'].value_counts())
```

## **Unsupervised Method: Hierarchical Clustering(Optional)**

```python
# Why this? It helps understand how data points form clusters at different
# distance thresholds it doesn’t require a fixed K like KMeans
# Hierarchical reveals segment splits at different levels.
import scipy.cluster.hierarchy as sch
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
# Create linkage matrix
linkage_matrix = linkage(X_scaled, method='ward')
```

```python
# Plot dendrogram
plt.figure(figsize=(12, 6))
dendrogram(linkage_matrix, truncate_mode='level', p=5)
plt.title("Hierarchical Clustering Dendrogram (truncated at 5 levels)")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
from scipy.cluster.hierarchy import fcluster
# Extract flat cluster labels by cutting the dendrogram at a distance threshold
hier_labels = fcluster(linkage_matrix, t=10, criterion='distance')
features_df['hier_cluster'] = hier_labels
```

## Hypothesis Testing: Segment Differences in Revenue & Discount Usage



```python
from scipy.stats import f_oneway

# Optional: Drop missing values for cleaner testing
df = features_df.dropna(subset=['total_revenue', 'flight_discount_proportion', 'segment'])

# Group data by segments
segments = df['segment'].unique()

# ANOVA for total_revenue
revenue_groups = [df[df['segment'] == seg]['total_revenue'] for seg in segments]
revenue_stat, revenue_p = f_oneway(*revenue_groups)

# ANOVA for flight_discount
discount_groups = [df[df['segment'] == seg]['flight_discount_proportion'] for seg in segments]
discount_stat, discount_p = f_oneway(*discount_groups)


print("=== ANOVA: Total Revenue Across Segments ===")
print(f"F-statistic: {revenue_stat:.2f}, p-value: {revenue_p:.4f}")
if revenue_p < 0.05:
    print("Statistically significant difference in total revenue between segments.")
else:
    print("No significant difference in total revenue between segments.")

print("\n=== ANOVA: Discount Usage Across Segments ===")
print(f"F-statistic: {discount_stat:.2f}, p-value: {discount_p:.4f}")
if discount_p < 0.05:
    print("Statistically significant difference in discount usage between segments.")
else:
    print("No significant difference in discount usage between segments.")
```
